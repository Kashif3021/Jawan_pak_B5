```python
# Define a list of dictionaries to store the quiz questions, options, and the correct answer
quiz_questions = [
    {
        "question": "How many Provinces in Pakistan?",
        "options": ["A) 1", "B) 2", "C) 3", "D) 4"],
        "correct": "D"
    },
    {
        "question": "How many Days in a week?",
        "options": ["A) 1", "B) 2", "C) 7", "D) 5"],
        "correct": "C"
    },
    {
        "question": "How many Months in a Year?",
        "options": ["A) 12", "B) 10", "C) 7", "D) 15"],
        "correct": "A"
    },
    {
        "question": "How many Bananas in a Dozen?",
        "options": ["A) 11", "B) 12", "C) 5", "D) 15"],
        "correct": "B"
    },
    {
        "question": "How many Eggs in a Half Dozen?",
        "options": ["A) 11", "B) 12", "C) 6", "D) 15"],
        "correct": "C"
    }
]

score = 0
total_questions = len(quiz_questions)

# Ask questions one by one
for i, item in enumerate(quiz_questions, start=1):
    print(f"\nQuestion {i}: {item['question']}")
    for option in item['options']:
        print(option)
    
# Get user input, trim whitespace, and convert to uppercase
user_answer = input("Enter your answer (A, B, C, or D): ").strip().upper()
    
# Check if the answer is correct
if user_answer == item['correct']:
        print("Correct!")
        score += 1
    else:
        print(f"Wrong! The correct answer was {item['correct']}.")

# At the end, show the total score
print(f"\nQuiz Finished!")
print(f"You got {score} out of {total_questions} correct!")

# Show performance feedback
if score == total_questions:
    print("Excellent!")
elif score >= total_questions / 2:
    print("Good job!")
else:
    print("Try again!")
```
