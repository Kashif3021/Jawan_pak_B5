Question 1:

Write a Python program that asks the user to guess a secret word.

Keep asking until the user enters the correct word ("python").

When guessed correctly, print "🎉 You guessed it!"

Expected Output:

Guess the word: java

Guess the word: ruby

Guess the word: python

🎉 You guessed it!

```python
# Set the secret word
secret_word = "python"

# Start an infinite loop that keeps asking until broken
while True:
    # Ask the user for their guess
    guess = input("Guess the word: ")
    
    # Check if the guess matches the secret word (ignoring capitalization)
    if guess.lower() == secret_word:
        print("🎉 You guessed it!")
        break  # Exit the loop since they guessed correctly
```

Q2 Stop sending emails after 3 users (Using for loop and break)

users = ["Ali", "Sara", "John", "Zara", "Mike"]

Output:

Email sent to Ali

Email sent to Sara

Email sent to John

Limit reached. Stop sending emails.

```python
users = ["Ali", "Sara", "John", "Zara", "Mike"]

# Initialize a counter to track how many emails have been sent
emails_sent = 0

for user in users:
    # Check if we have already sent 3 emails
    if emails_sent == 3:
        print("Limit reached. Stop sending emails.")
        break  # Exit the loop immediately
        
    print(f"Email sent to {user}")
    emails_sent += 1  # Increment the counter
```

Q3. Stop monitoring temperature if critical temperature reached (Using for loop and break)

temps = [22, 25, 30, 45, 28] critical = 40

Ouput:

Temperature: 22

Temperature: 25

Temperature: 30

Critical temperature reached! Stop monitoring.

```python
temps = [22, 25, 30, 45, 28]
critical = 40

for temp in temps:
    # Check if the temperature has hit or exceeded the critical limit
    if temp >= critical:
        print("Critical temperature reached! Stop monitoring.")
        break  # Exit the loop immediately
        
    print(f"Temperature: {temp}")
```

Q4. Calculate total price of items in a shopping cart (Using For loop)

cart = [100, 250, 50, 300]

Output:

Total Price: 700

```python
cart = [100, 250, 50, 300]

# Initialize a variable to store the running total
total_price = 0

# Loop through each item price in the cart
for price in cart:
    total_price += price  # Add the current item's price to the total

# Print the final total
print(f"Total Price: {total_price}")
```

Q5. Skip customers with unpaid bills (Using For loop and continue)

customers = [ {"name": "Ali", "paid": True},

{"name": "Sara", "paid": False},

{"name": "John", "paid": True}
]

Output:

Access granted to: Ali

Access granted to: John

```python
customers = [
    {"name": "Ali", "paid": True},
    {"name": "Sara", "paid": False},
    {"name": "John", "paid": True}
]

for customer in customers:
    # If the customer has not paid, skip the rest of the loop for them
    if not customer["paid"]:
        continue
        
    print(f"Access granted to: {customer['name']}")
```

users = ["Ali", "Sara", "John"]

Hello John, your subscription is active!

Q7. Extract domain names from email list (Using for loop)

emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

Output:

domains: ['gmail.com', 'yahoo.com', 'hotmail.com']



```python
emails = ["ali@gmail.com", "sara@yahoo.com", "john@hotmail.com"]

# Initialize an empty list to store the extracted domains
domains = []

for email in emails:
    # Split the email string at the '@' character and take the second part [1]
    domain = email.split("@")[1]
    domains.append(domain)

print(f"domains: {domains}")
```

Q8. Print only overdue tasks (Using for loop)

tasks = [ {"task": "Pay bills", "status": "done"},

{"task": "Submit report", "status": "pending"},

{"task": "Clean house", "status": "pending"}
]

Ouput:

Overdue: Submit report

Overdue: Clean house

```python
tasks = [ 
    {"task": "Pay bills", "status": "done"},
    {"task": "Submit report", "status": "pending"},
    {"task": "Clean house", "status": "pending"} 
]

for item in tasks:
    # Check if the status of the current task is 'pending'
    if item["status"] == "pending":
        print(f"Overdue: {item['task']}")
```

Q9:

Write a Python program that keeps asking the user to enter numbers and prints them.

The program should stop the loop if the user enters a negative number.

Use a while True loop with a break statement

Expected Output:

Enter a number (negative to stop): 5

You entered: 5

Enter a number (negative to stop): 10

You entered: 10

Enter a number (negative to stop): -3

Negative number entered. Loop stopped!

```python
##### Start an infinite loop
while True:
    # Ask the user for input and convert it to an integer
    number = int(input("Enter a number (negative to stop): "))
    
    # Check if the number is negative (less than 0)
    if number < 0:
        print("Negative number entered. Loop stopped!")
        break  # Exit the loop immediately
        
    print(f"You entered: {number}")
```

Q10. Skip weekends while printing workdays (Using for loop and continue)

days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

Output:

Workday: Monday

Workday: Tuesday

Workday: Wednesday

Workday: Thursday

Workday: Friday

```python
days = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

for day in days:
    # If the day is Saturday or Sunday, skip the rest of the loop
    if day == "Saturday" or day == "Sunday":
        continue
        
    print(f"Workday: {day}")
```
