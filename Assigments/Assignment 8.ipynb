Make a class called Employee. The __init__() method for Employee should store its different attributes: name, number, address, salary, post, year of joining, education, and his year of experience.

Make a method called describe_employee() that prints the detail Information about employee, a method update_ year_of_experience(), and a method called increase_salary() that do increamnet in it's salary.

Make an instance called employee101 from your class. Print the four attri-
butes individually, and then call all it's methods for describing employee, ssalary increamnet and increase year of experience.

Create 10 different instances from the class, and call describe_employee() for each
instance. Then do some updations with it's methods.



```python
class Employee:
    def __init__(self, name, number, address, salary, post, year_of_joining, education, year_of_experience):
        """Initialize employee attributes."""
        self.name = name
        self.number = number
        self.address = address
        self.salary = salary
        self.post = post
        self.year_of_joining = year_of_joining
        self.education = education
        self.year_of_experience = year_of_experience

    def describe_employee(self):
        """Print detailed information about the employee."""
        print(f"--- Employee Details: {self.name} ---")
        print(f"Contact Number:     {self.number}")
        print(f"Address:            {self.address}")
        print(f"Designation/Post:   {self.post}")
        print(f"Education:          {self.education}")
        print(f"Year of Joining:    {self.year_of_joining}")
        print(f"Experience:         {self.year_of_experience} years")
        print(f"Current Salary:     ${self.salary:,.2f}")
        print("-" * 35)

    def update_year_of_experience(self, new_experience):
        """Set or increment years of experience."""
        self.year_of_experience = new_experience
        print(f"Updated experience for {self.name}: {self.year_of_experience} years.")

    def increase_salary(self, increment_amount):
        """Increase the employee's salary by a specified amount."""
        self.salary += increment_amount
        print(f"Salary increased for {self.name} by ${increment_amount:,.2f}. New Salary: ${self.salary:,.2f}")


# ==========================================
# Part 1: Instance employee101 & Demonstration
# ==========================================

print("=== DEMONSTRATING EMPLOYEE 101 ===")
employee101 = Employee(
    name="Alice Johnson",
    number="555-0192",
    address="123 Elm St, Springfield",
    salary=75000,
    post="Senior Software Engineer",
    year_of_joining=2020,
    education="B.S. Computer Science",
    year_of_experience=5
)

# Printing four attributes individually
print(f"Employee Name:       {employee101.name}")
print(f"Employee Post:       {employee101.post}")
print(f"Employee Salary:     ${employee101.salary}")
print(f"Employee Education:  {employee101.education}\n")

# Calling methods on employee101
employee101.describe_employee()
employee101.update_year_of_experience(6)
employee101.increase_salary(5000)
employee101.describe_employee()


# ==========================================
# Part 2: Creating 10 Employee Instances
# ==========================================

employees = [
    Employee("Bob Smith", "555-0101", "456 Oak Ave", 50000, "Junior Developer", 2023, "B.S. IT", 1),
    Employee("Charlie Brown", "555-0102", "789 Pine Rd", 65000, "QA Analyst", 2021, "B.S. Software Eng", 3),
    Employee("Diana Prince", "555-0103", "321 Maple St", 95000, "Project Manager", 2018, "M.B.A.", 8),
    Employee("Evan Wright", "555-0104", "654 Cedar Ln", 55000, "UI/UX Designer", 2022, "B.A. Graphic Design", 2),
    Employee("Fiona Gallagher", "555-0105", "987 Birch Dr", 110000, "DevOps Lead", 2017, "B.S. Computer Eng", 9),
    Employee("George Clark", "555-0106", "147 Spruce St", 48000, "Support Specialist", 2024, "A.S. Information Systems", 1),
    Employee("Hannah Abbott", "555-0107", "258 Ash Ave", 82000, "Data Scientist", 2020, "M.S. Data Analytics", 4),
    Employee("Ian Malcolm", "555-0108", "369 Walnut Rd", 105000, "System Architect", 2016, "Ph.D. Computer Science", 10),
    Employee("Julia Roberts", "555-0109", "741 Chestnut St", 70000, "HR Manager", 2019, "B.A. Human Resources", 6),
    Employee("Kevin Bacon", "555-0110", "852 Poplar Ln", 60000, "Database Admin", 2022, "B.S. Computer Science", 3)
]

print("\n=== DESCRIBING ALL 10 EMPLOYEES ===")
for emp in employees:
    emp.describe_employee()


# ==========================================
# Part 3: Performing Updates
# ==========================================

print("\n=== PERFORMING UPDATES ON EMPLOYEES ===")
# Update experience for Bob (index 0) and increase Diana's salary (index 2)
employees[0].update_year_of_experience(2)
employees[0].increase_salary(4000)

employees[2].increase_salary(12000)
employees[2].update_year_of_experience(9)

# Verify updates with describe_employee()
print("\n=== VERIFYING UPDATED EMPLOYEES ===")
employees[0].describe_employee()
employees[2].describe_employee()
```

```python
# ==========================================
# Helper Functions (Replacing Class Methods)
# ==========================================

def create_employee(name, number, address, salary, post, year_of_joining, education, year_of_experience):
    """Factory function to construct an employee dictionary."""
    return {
        "name": name,
        "number": number,
        "address": address,
        "salary": salary,
        "post": post,
        "year_of_joining": year_of_joining,
        "education": education,
        "year_of_experience": year_of_experience
    }

def describe_employee(emp):
    """Print detailed information about the employee dictionary."""
    print(f"--- Employee Details: {emp['name']} ---")
    print(f"Contact Number:     {emp['number']}")
    print(f"Address:            {emp['address']}")
    print(f"Designation/Post:   {emp['post']}")
    print(f"Education:          {emp['education']}")
    print(f"Year of Joining:    {emp['year_of_joining']}")
    print(f"Experience:         {emp['year_of_experience']} years")
    print(f"Current Salary:     ${emp['salary']:,.2f}")
    print("-" * 35)

def update_year_of_experience(emp, new_experience):
    """Update years of experience in the employee dictionary."""
    emp["year_of_experience"] = new_experience
    print(f"Updated experience for {emp['name']}: {emp['year_of_experience']} years.")

def increase_salary(emp, increment_amount):
    """Increase salary in the employee dictionary."""
    emp["salary"] += increment_amount
    print(f"Salary increased for {emp['name']} by ${increment_amount:,.2f}. New Salary: ${emp['salary']:,.2f}")


# ==========================================
# Part 1: Individual Employee employee101
# ==========================================

print("=== DEMONSTRATING EMPLOYEE 101 ===")
employee101 = create_employee(
    name="Alice Johnson",
    number="555-0192",
    address="123 Elm St, Springfield",
    salary=75000,
    post="Senior Software Engineer",
    year_of_joining=2020,
    education="B.S. Computer Science",
    year_of_experience=5
)

# Printing four attributes individually from the dictionary
print(f"Employee Name:       {employee101['name']}")
print(f"Employee Post:       {employee101['post']}")
print(f"Employee Salary:     ${employee101['salary']}")
print(f"Employee Education:  {employee101['education']}\n")

# Calling functions on employee101
describe_employee(employee101)
update_year_of_experience(employee101, 6)
increase_salary(employee101, 5000)
describe_employee(employee101)


# ==========================================
# Part 2: Creating 10 Employee Dictionaries
# ==========================================

employees = [
    create_employee("Bob Smith", "555-0101", "456 Oak Ave", 50000, "Junior Developer", 2023, "B.S. IT", 1),
    create_employee("Charlie Brown", "555-0102", "789 Pine Rd", 65000, "QA Analyst", 2021, "B.S. Software Eng", 3),
    create_employee("Diana Prince", "555-0103", "321 Maple St", 95000, "Project Manager", 2018, "M.B.A.", 8),
    create_employee("Evan Wright", "555-0104", "654 Cedar Ln", 55000, "UI/UX Designer", 2022, "B.A. Graphic Design", 2),
    create_employee("Fiona Gallagher", "555-0105", "987 Birch Dr", 110000, "DevOps Lead", 2017, "B.S. Computer Eng", 9),
    create_employee("George Clark", "555-0106", "147 Spruce St", 48000, "Support Specialist", 2024, "A.S. Information Systems", 1),
    create_employee("Hannah Abbott", "555-0107", "258 Ash Ave", 82000, "Data Scientist", 2020, "M.S. Data Analytics", 4),
    create_employee("Ian Malcolm", "555-0108", "369 Walnut Rd", 105000, "System Architect", 2016, "Ph.D. Computer Science", 10),
    create_employee("Julia Roberts", "555-0109", "741 Chestnut St", 70000, "HR Manager", 2019, "B.A. Human Resources", 6),
    create_employee("Kevin Bacon", "555-0110", "852 Poplar Ln", 60000, "Database Admin", 2022, "B.S. Computer Science", 3)
]

print("\n=== DESCRIBING ALL 10 EMPLOYEES ===")
for emp in employees:
    describe_employee(emp)


# ==========================================
# Part 3: Performing Updates
# ==========================================

print("\n=== PERFORMING UPDATES ON EMPLOYEES ===")
# Updating Bob (index 0)
update_year_of_experience(employees[0], 2)
increase_salary(employees[0], 4000)

# Updating Diana (index 2)
increase_salary(employees[2], 12000)
update_year_of_experience(employees[2], 9)

# Verify updates
print("\n=== VERIFYING UPDATED EMPLOYEES ===")
describe_employee(employees[0])
describe_employee(employees[2])
```
