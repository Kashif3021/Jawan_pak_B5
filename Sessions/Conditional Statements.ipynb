### Conditional Statements in Python
#### Conditional statements allow you to execute code based on condition 
#### evaluates to True or False. They are essential for controlling the flow #### of a program and making decisions based on different inputs or conditions 


### Type of Conditonal Statements

### 'if' Conditional Statement

#### Syntax:
#### if condition:
