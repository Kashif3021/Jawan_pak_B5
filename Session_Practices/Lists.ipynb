## Slicing

```python
student_marks = [40,60,30,40,30,10,70,40,30,10]
print(student_marks)
```

```python
# we are accessing to range of element in positive 
```

```python
print(student_marks[1:4])
```

```python
# IN case w can give start position not end position 
```

```python
print(student_marks[2:])
```

```python
print(student_marks[:7])
```
