## Dictionary
In Python, a dictionary is mutable collection of key value pairs. It is one of the most versatile data structures in Python, 
allowing you to efficiently store and retrieve data based on custom keys.
#### Creating a Dictionary
You define a dictionary using curly bracket {} with keys and values separated by colons.
##### my_dic = {key1 : Value1,Key2 : Value2,...} 
##### Key: Must be unique and immutable (strings, numbers, or tuples).
##### Value: Can be any data type and does not need to be unique.

```python
My_bio_data = {
    "Name":'Kashif',"age":26,"job":"Data Analyst"
}
print(My_bio_data)
```

#### Accessing Values
You can grab a value by passing its key inside square brackets [], 
#### or by using the .get() method 
#### (which won't crash your program if the key doesn't exist).
#### my_dic.get(key)
#### or my_dic[Key]

```python
# Accessing Mannualy 
My_bio_data["Name"]
```

```python
# Accessing through Method
My_bio_data.get("Name")
```

### Nested Dictionary
#### A nested dictionary is simply a dictionary inside another dictionary.
#### Think of it like a folder on your computer that contains subfolders instead of just plain files.

```python
# Nested Dictionaries
employee = {
    101 : {'name': 'khan', 'age' : 22, 'Salary' : 2222222},
    102 : {'name': 'alice', 'age' : 33, 'Salary' : 2222},
    103 : {'name': 'john cena', 'age' : 11, 'Salary' : 22}
}
employee
```

```python
# Print of  khan age 
print(employee.get("age"))
```

```python
# Print of  khan age 
print("Khan age :",employee.get(101).get("age"))
```

```python
# Mannual prodesso
print(employee[101])
```

```python
# Mannual prodesso
print(employee[101]["age"])
```

```python
 # Mannual processing updating age
employee[101]["age"] = 29
```
