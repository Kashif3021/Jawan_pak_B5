## Tuple unpacking (sometimes called sequence unpacking)
### Tuple Unpacking is one of Python’s coolest shortcut features. It allows you to extract elements from a tuple and assign them to individual variables in a single, clean line of code.

### Instead of pulling items out one by one using their index numbers,
### you let Python unpack the whole container at once.

```python
user_data = ("Kashif","Night Shift","Data Analyst")
user_data
```

```python
#Unpacking
Name,Shift,Role = user_data
```

```python
Name
```

```python
Shift
```

```python
Role
```

```python
# or we can seperate below case
user_data = ("Kashif","Night Shift","Data Analyst")
user_data
```

```python
# You have used tuple unpacking with the star (*) operator! This is a fantastic, 
# highly efficient Python feature often called extended unpacking
Name,*info = user_data
```

```python
info
```

```python
# or we can seperate below case
user_data = ("Kashif","Night Shift","Data Analyst")
```

```python
*info,role = user_data
```

```python
role
```

```python
info
```
