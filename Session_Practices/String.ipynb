### Types of String Formatting
A formatted string in Python is a way to insert variables or expressions inside a string. It allows you to format the output in a readable and controlled way.
There are multiple ways to format strings in Python:
#### 1.	Old-style formatting (% operator)
#### 2.	str.format() method
#### 3.	F-strings (formatted string literals)


### Formatted String - % Operator
#### Old-style formatting (% operator)
#### This approach uses the % operator and is similar to string formatting in languages like C.
#### Syntax: "string % value"

```python
Student = "I am student of Bacholar of Science"
```

```python
My_age = 26
```

```python
print(" Who are You ? %s and what is your %d" % (Student, My_age))
```

### Formatted String - str.format()
#### str.format() method
In Python 3, the format() method is more powerful and flexible than the old-style %
formatting.

#### Syntax: "string {}".format(value)

```python
Name = "Kashif Ali"
age = 26
```

```python
print("What is your Name : {} and your age {}".format(Name, age))
```

```python
print("What is your Name : {Name} and your age {age}".format(Name = "Kashif", age = 29 ))
```

### Formatted String – F-strings
#### F-strings (formatted string literals)
In Python 3.6, F-strings are the most concise and efficient way to format strings. You
prefix the string with an f or F, and variables or expressions are embedded directly within curly braces {}.
#### Syntax: f"string {variable}"

```python
Name = "Kashif Ali"
age = 26
```

```python
print(f"what is your Name {Name} and your age {age}")
```

```python
a = 29 
b = 60
```

```python
print(f"The sum of \"a\" = {a}, and \"b\" = {b} is = {a+b}") # note we can perofrome operation in the curely braces on variable
```

### Escape Characters
Escape characters in Python are special characters used in strings to represent
#### whitespace, symbols, or control characters that would otherwise be difficult to include. 
##### An escape character is a backslash \ followed by the character you want to insert.


```python
print("My name is \"kashif\"") # \"variable\" is used for printing Bouble quotes marks
```

```python
print("My name is kashif \nand\n \nMy age is 29\n") #\n variable \n is used for new line
```

```python
print("My name is\tMuhammad kashif\t") # \t variable\t is used for tab
```

```python
print("My name is \\Kashif\\")
```

#### String Slicing
Slicing in Python is a feature that enables accessing parts of the sequence. String slicing allows you to get subset of characters from a string using a specified range of indices.
 
#### Syntax:
#### string[start : end : step]
#### start : The index to start slicing (inclusive). Default value is 0.
#### end : The index to stop slicing (exclusive). Default value is length of string.
#### Step:How much to increment the index after each character. Default value is 1. 


```python
Name
```

```python
Name[0:9]
```

```python
Name[1:]
```
