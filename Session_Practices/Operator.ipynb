### Operators in Python
Operators in Python are special symbols or keywords used to perform operations on operands (variables and values).
#### Operators: These are the special symbols/keywords. Eg: + , * , /, etc.
#### Operand: It is the value on which the operator is applied.


### Types of Operators


#### Arithmetic Operators
Arithmetic operators are used with numeric values to perform mathematical operations such as addition, subtraction, multiplication, and division 


```python
A = 10
B = 20
```

```python
# addition
Sum = A + B
Sum
```

```python
# Substraction
Sub = A - B
Sub
```

```python
# Multiplication
Mut = A*B
Mut
```

```python
# Divison
divison = A/B
divison
```

```python
mod = A%B
mod
```

```python
fd = A//B
fd
```

```python
ex = A**B
ex
```

### Logical Operators 
Logical operators are used to combine conditional statements


### and
#### Returns True if both statements are true
#### a > 2 and a < 5


```python
# when both Statements are false then
40 > 60 and 80 < 40
```

```python
# when both Statements are True then
40 > 30 and 80 < 90
```

```python
# when anyone Statement is True then
40 > 60 and 80 < 100
```

### or
#### Returns True if one of the statements is true
#### a > 5 or a < 10


```python
# when anyone Statement is True then
40 > 60 or 80 < 100
```

```python
# when Both Statements are false then
40 > 60 or 80 < 70
```

```python
# when Anyone Statement is True then
50 > 40 or 80 < 100
```

```python
# when Both Statements are True then
50 > 40 or 80 < 100
```

### not
#### Reverse the result, returns False if the result is true
#### not(a > 2 and a < 10)


```python
not (40 > 30) or (30 < 20 and 40 == 40)
```

```python
not (40 > 30 and 40 == 40)
```

```python
not (40 > 50) 
```

### Membership
Membership operators checks whether a given value is a member of a sequence (such as strings, lists, and tuples) or not.


#### in Operator
in

```python
text = "Java"
```

```python
"J" in text
```

```python
"z" in text
```

```python
student = ["shams","Usama","Azhar"]
```

```python
"Nasar" in student
```
