### Loop Control Statements
Loop control statements allow you to alter the normal flow of a loop.


#### break Statement
break Statement: The break statement terminates the loop entirely, exiting from it
##### immediately. or
statement is used to exit a loop prematurely when a specific condition is met.


```python
for i in range(5):
if i == 3:
break
print(i)  # Output: 0  1  2
Above example, the loop terminated when condition met true for i == 3
```

```python
list1 = [1,2,3,4,5,7]
for i in list1:
    if i==4:
        break
    print(i)
```

### continue Statement
statement is used to skip the current iteration of a loop when a specific condition is met, and it continues with the next iteration of the loop

```python
list1 = [1,2,3,4,5,7]
for i in list1:
    if i==4:
        continue
    print(i)
```

```python
for i in range(5):
    if i == 3:
        continue
    print(i)  # Output: 0  1  2
# Above example, the loop terminated when condition met true for i == 3
```
