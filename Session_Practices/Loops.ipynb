## ***Why do we use for loop in python?***

***A for loop in Python is used to iterate over a sequence (like a list, tuple, string, or range) and execute a block of code repeatedly for each item in the sequence. It helps automate repetitive tasks, making the code more concise and efficient***
### Syntax:
for variable in sequence:


```python
my_tuple = (1,2,3,4,6,9)
```

```python
for print_one_by_one in my_tuple:
    print(print_one_by_one)
```

```python
# if we want to count the number of item then we use enumerator()
for index,fruit in enumerate(Fruit_list,):
    print(index,fruit)
```

```python
# if we want to count the number of item then we use enumerator()
# if we want to start the count number from one then put stat=1  
for index,fruit in enumerate(Fruit_list,start=1):
    print(index,fruit)
```

### Using range() Function
#### To repeat a block of code a specified number of times, we use the range() function.
#### The range() function returns a sequence of numbers, starting from 0 by default,
#### increments by 1 (by default), and stops before a specified number 


### Syntax:
#### range(stop)
#### range(start, stop)
#### range(start, stop, step)
#### • start: (optional) The beginning of the sequence. Defaults is 0. (inclusive)
#### • stop: The end of the sequence (exclusive).
#### • step: (optional) The difference between each number in the sequence.
Defaults is 1.


```python
for x in range(10):
    print(x)
```

```python
for x in range(2,50,5):
    print(x)
```

```python
for x in range(1,10):
    print(f"2*{x}= {2*x}")
```

### Nested Loops in Python
Loop inside another loop is nested loop. This means that for every single time the outer loop runs, the inner loop runs all of its iterations.
##### Why Use Nested Loops?
·       Handling Multi-Dimensional Data: Such as matrices, grids, or lists of lists.
·       Complex Iterations: Operations depend on multiple variables or dimensions.
·       Pattern Generation: Creating patterns, such as in graphics or games.
### Nested Loop Syntax
##### Syntax:
Outer_loop:
#### 	inner_loop:
####     	  # Code block to execute - inner loop
####    Code block to execute - outer loop


```python
for x in range(5): # out loops run one time
    for num in range(1,5): # it run all  iteration 
        print(num)
```

```python
for x in range(3): # this inner loops 
    print("Running outer loops",x)
    for y in range(1,5):
        print(y)
```

```python
student_name = ["shams","Alia","Azra","Umar"]
visted_city = ["Islamabad","Peshawar","Karachi"]
```

```python
for student in student_name:
    for visit in visted_city:
        print(student,visit)
```

```python
print(student_name)
print(visted_city)
```

```python
for student in student_name:
    print(student,"visited the following city")
    for visit in visted_city:
        print(visit)
```

```python
for student in student_name:
    print(student,"visited the following city")
    for index, visit in enumerate(visted_city,start=1):
        print(index,visit)
    print('')
```

### While Loop
#### The while loop repeatedly executes a block of code as long as a given condition remains
#### True. It checks the condition before each iteration.
### Syntax:
#### while condition: 
#### Code block to execute


```python
num = 1
while num <= 10:
    print(num)
    num+=1
```

```python
cities = []
count = ''
```

```python
cities = []
count = ''

while count != 'quit':
    city = input('enter a city')
    if city == 'quit':
        count = 'quit'
    else:
        cities.append(city)
```

```python
print(cities)
print(count)
```
